# 04_Data Cleaning

```python
df_clean = df.copy()

df_clean = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["Sales"] >= 0) &
    (df_clean["OrderValue"] >= 0)
].copy()
```

```python
# Converting date columns to datetime format

df_clean["OrderDate"] = pd.to_datetime(
    df_clean["OrderDate"],
    format="%d/%m/%Y"
)

df_clean["SignupDate"] = pd.to_datetime(
    df_clean["SignupDate"],
    format="%d/%m/%Y"
)
```

```python
# Saving the cleaned dataset to a new CSV file
df_clean.to_csv("../data/E-Commerce_Sales_Cleaned.csv", index=False)
```

# 06_Feature Engineering

```python
# Creating new columns for Year, Month, MonthName, Quarter, and YearMonth

df_clean["Year"] = df_clean["OrderDate"].dt.year
df_clean["Month"] = df_clean["OrderDate"].dt.month
df_clean["MonthName"] = df_clean["OrderDate"].dt.month_name()
df_clean["Quarter"] = "Q" + df_clean["OrderDate"].dt.quarter.astype(str)
df_clean["YearMonth"] = df_clean["OrderDate"].dt.to_period("M")
```

```python
# Creating a new column for CustomerType based on the number of orders per customer

orders_per_customer = df_clean.groupby("CustomerID").size()
df_clean["OrderCount"] = df_clean["CustomerID"].map(orders_per_customer)

df_clean["CustomerType"] = np.where(
    df_clean["OrderCount"] == 1,
    "New Customer",
    "Repeat Customer"
)

df_clean["CustomerType"].value_counts()
```

# 07_Customer Analysis

```python
# Creating a customer summary DataFrame with aggregated metrics

customer_summary = (
    df_clean.groupby("CustomerID")
    .agg(
        OrderCount=("OrderID", "count"),
        TotalSales=("Sales", "sum"),
        AvgOrderValue=("Sales", "mean")
    )
    .reset_index()
)

customer_summary["CustomerType"] = np.where(
    customer_summary["OrderCount"] == 1,
    "New Customer",
    "Repeat Customer"
)
```

```python
customer_summary["CustomerType"].value_counts() # Count of each customer type
```

```python
# Creating value segments based on total sales

customer_summary["CustomerValueSegment"] = pd.cut(
    customer_summary["TotalSales"],
    bins=[0, 166, 290, 465, float("inf")],
    labels=["Low Value", "Medium-Low", "Medium-High", "High Value"],
    include_lowest=True
)

customer_summary["CustomerValueSegment"].value_counts()
```

```python
# Grouping by CustomerValueSegment and calculating the mean of OrderCount, TotalSales, and AvgOrderValue

customer_summary.groupby(
    "CustomerValueSegment",
    observed=True
)[
    ["OrderCount", "TotalSales", "AvgOrderValue"]
].mean()
```

```python
# Grouping by CustomerSegment and calculating the number of customers, total sales, average order value, and number of orders

customer_segment_analysis = (
    df_clean.groupby("CustomerSegment")
    .agg(
        Customers=("CustomerID", "nunique"),
        TotalSales=("Sales", "sum"),
        AvgOrderValue=("Sales", "mean"),
        Orders=("OrderID", "count")
    )
    .sort_values("TotalSales", ascending=False)
)

customer_segment_analysis
```

# 08_Sales Trend Analysis

```python
# Grouping by YearMonth and calculating total sales, number of orders, and number of unique customers

monthly_sales = (
    df_clean.groupby("YearMonth")
    .agg(
        TotalSales=("Sales", "sum"),
        Orders=("OrderID", "count"),
        Customers=("CustomerID", "nunique")
    )
    .reset_index()
)

monthly_sales["MoM_Growth"] = monthly_sales["TotalSales"].pct_change() * 100
monthly_sales
```

```python
# Find the months with the highest and lowest sales

max_sales = monthly_sales.loc[monthly_sales["TotalSales"].idxmax()]
min_sales = monthly_sales.loc[monthly_sales["TotalSales"].idxmin()]

max_sales, min_sales
```

# 09_Category Analysis

```python
# Grouping by Category and calculating total sales, number of orders, number of unique customers, and average order value

category_analysis = (
    df_clean.groupby("Category")
    .agg(
        TotalSales=("Sales", "sum"),
        Orders=("OrderID", "count"),
        Customers=("CustomerID", "nunique"),
        AvgOrderValue=("Sales", "mean")
    )
    .reset_index()
    .sort_values("TotalSales", ascending=False)
)

category_analysis
```

# 10_Product Analysis

```python
# Grouping by ProductID and ProductName and calculating total sales, number of orders, and quantity sold

product_analysis = (
    df_clean.groupby(["ProductID", "ProductName"])
    .agg(
        TotalSales=("Sales", "sum"),
        Orders=("OrderID", "count"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("TotalSales", ascending=False)
)

product_analysis.head(10)
```

# 11_Key Insights
  ### Sales Performance

  - May 2025 recorded the highest monthly sales at $130.6K.
  - February 2026 recorded the lowest monthly sales at $75.1K.
  - Sales declined 42.45% MoM in February 2026, followed by a partial recovery in March.

  ### Customer Insights

  - Repeat customers represent the majority of observed customers.
  - High Value customers have higher average order frequency and spending than lower-value customer segments.
  - Regular customers generate the highest total sales among the dataset's predefined customer segments.

  ### Category Insights

  - Electronics is the largest sales-generating category.
  - Accessories has the highest order volume but a relatively low average order value.
  - Home Office has the highest average order value despite lower order volume.

  ### Product Insights

  - Headphones is the top product by total sales.
  - Office Chair generates nearly the same sales as Headphones with fewer orders, indicating higher sales value per order.

# 12_Data Limitations

- The dataset covers January 2024 to June 2026.
- 2026 contains only six months of data, so full-year comparisons with 2024 and 2025 should be interpreted with caution.
- CustomerValueSegment is based on quartiles of observed customer sales, so the segment thresholds are relative to this dataset rather than business-defined thresholds.
- The dataset does not contain a Profit field, so profitability analysis is outside the scope of this project.
